Halaman ini ditulis menggunakan Jupyter Notebook (`.ipynb`). Jupyter Book akan merender kode Python beserta outputnya secara otomatis di halaman web statis Anda.

```python
import numpy as np

# Membuat dua matriks 2x2
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matriks A:\n", A)
print("\nMatriks B:\n", B)
```

```python
# Melakukan perkalian matriks
C = np.dot(A, B)
print("Hasil Perkalian Matriks (A x B):\n", C)
```
